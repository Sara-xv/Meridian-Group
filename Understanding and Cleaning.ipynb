### Combine datasets

```python
perf_summary = (
    P.groupby("employee_id")
    .agg(
        review_count=("performance_id", "count"),
        avg_performance_score=("performance_score_1to5", "mean"),
        avg_manager_rating=("manager_rating_1to5", "mean"),
        avg_goal_completion_rate=("goal_completion_rate_pct", "mean"),
        trained_review_periods=("had_recent_training_last_180d", "sum"),
        last_review_date=("review_date", "max"),
    )
    .reset_index()
)
```

```python
perf_summary.sample(10)
```

```python
te_completed = TE[TE["completion_status"] == "Completed"]
 
training_summary = (
    TE.groupby("employee_id")
    .agg(total_enrollments=("enrollment_id", "count"))
    .reset_index()
)
training_summary["completed_count"] = (
    TE.groupby("employee_id")["completion_status"]
    .apply(lambda s: (s == "Completed").sum())
    .values
)
completed_stats = (
    te_completed.groupby("employee_id")
    .agg(
        avg_assessment_score=("assessment_score", "mean"),
        avg_satisfaction=("participant_satisfaction_1to5", "mean"),
    )
    .reset_index()
)
training_summary = training_summary.merge(completed_stats, on="employee_id", how="left")
```

```python
training_summary.sample(10)
```

```python
te_with_program_info = TE.merge(
    TP[["program_id", "category", "cost_per_participant_usd"]],
    on="program_id", how="left"
)

cost_summary = (
    te_with_program_info[te_with_program_info["completion_status"] == "Completed"]
    .groupby("employee_id")["cost_per_participant_usd"]
    .sum()
    .reset_index()
    .rename(columns={"cost_per_participant_usd": "total_training_cost_usd"})
)
```

```python
cost_summary.sample(10)
```

```python
category_flags = (
    te_with_program_info[te_with_program_info["completion_status"] == "Completed"]
    .assign(flag=1)
    .pivot_table(index="employee_id", columns="category", values="flag",
                 aggfunc="max", fill_value=0)
    .astype("int64")
)
category_flags.columns = [f"trained_in_{c.lower().replace(' ', '_')}" for c in category_flags.columns]
category_flags = category_flags.reset_index()
 
training_summary = training_summary.merge(cost_summary, on="employee_id", how="left")
training_summary = training_summary.merge(category_flags, on="employee_id", how="left")
```

```python
training_summary.sample(10)
```

```python
events_summary = (
    HR_E.groupby("employee_id")
    .agg(
        promotion_count=("event_type", lambda s: (s == "Promotion").sum()),
        salary_increase_count=("event_type", lambda s: (s == "Salary Increase").sum()),
        has_exited=("event_type", lambda s: int((s == "Exit").any())),
    )
    .reset_index()
)
```

```python
events_summary.sample(10)
```

```python
combined = E.merge(perf_summary, on="employee_id", how="left")
combined = combined.merge(training_summary, on="employee_id", how="left")
combined = combined.merge(events_summary, on="employee_id", how="left")
 

category_cols = [c for c in combined.columns if c.startswith("trained_in_")]
count_cols = (["total_enrollments", "completed_count", "promotion_count",
                "salary_increase_count", "has_exited", "review_count",
                "trained_review_periods", "total_training_cost_usd"] + category_cols)
for col in count_cols:
    if col in combined.columns:
        combined[col] = combined[col].fillna(0).astype("int64" if col != "total_training_cost_usd" else "float64")
 
print(f"Final combined dataset: {len(combined):,} rows (should exactly equal number of employees, i.e., "
      f"{len(E):,}) × {combined.shape[1]} columns")
assert len(combined) == len(E), "Final row count does not match number of employees - possible fan-out!"
 
print("\ndtypes before saving:")
print(combined.dtypes)
```

```python
combined.to_parquet("combined_employee_dataset.parquet", index=False, engine="pyarrow")
print("\n Save: combined_employee_dataset.parquet")
 
```
